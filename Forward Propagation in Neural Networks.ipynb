```python
def initialize_parameters():
    np.random.seed(1)
    W = np.random.randn(2, 1)*0.01
    b = np.zeros((1,1))
    return W,b
```

```python
def forward_parameters(X,W,b):
    Z =np.dot(X,W) + b
    A = 1/(1+ np.exp(-Z))
    return A
```

```python
W, b = initialize_parameters()
A = forward_parameters(X,W,b)
print("Final Output:", A)
```
