# LSTM (Long Short-term memory) in Deep learnig

## 1. What is LSTM?
LSTM stands for Long Short-Term Memory, a special type of Recurrent Neural Network (RNN) that learns long-term dependencies in sequential data.

**Example:** in a sentence, LSTM can remember the subject mentioned earlier to predict the correct mext wond

## 2. Why do we use LSTM?
LSTM helpt solve the vanishing gradient problem that makes tradional Nis struggle to learn long-term dependencies.

**Example:** Predicheng a movie review's sentiment using words from the entire review.

## 3. Four important components of LSTM

### 1. Forget Gate
Decides which information should be removed from the cell state.

**Example:** Ignoring an irrelevant word from an earlier sentence.

### 2. Input Gate
Decides which new information should be stored.

**Example:** Remembering a person's name in a conversation.

### 3. Cell State
Carries important information through the sequence

**Example:** Keeping track of the topic throughout a long conversation

### 4. Output Gate
Decides what information should be used for the current output.

**Example:** Keeping track of the topic throughout a long conversation

### Output Gate
Decides what information should be used for the current output

**Example:** Predicting the next word using relevant context.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Input data: 5 sequences, 3 time steps, 1 feature
X = np.array(
    [
        [[1], [2], [3]],
        [[2], [3], [4]],
        [[3], [4], [5]],
        [[4], [5], [6]],
        [[5], [6], [7]],
    ],
    dtype=np.float32,
)

# Target output (fixed closing parenthesis)
y = np.array([4, 5, 6, 7, 8], dtype=np.float32)

# Create model (fixed assignment operator =)
model = Sequential([LSTM(10, input_shape=(3, 1)), Dense(1)])

# Compile model
model.compile(optimizer="adam", loss="mse")

# Train model
model.fit(X, y, epochs=100, verbose=0)

# Predict output for [6, 7, 8]
test = np.array([[[6], [7], [8]]], dtype=np.float32)
prediction = model.predict(test, verbose=0)
print("Predicted value:", prediction[0][0])
```
