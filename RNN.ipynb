## 1. What is a Recurrent Neural Network (RNN)?
**Definition (English):** A Recurrent Neural Network (RNN) is a type of artificial neural network designed to process sequential data by using information from previous time steps to help process the current input.

**Examples of sequential data:**
- Text and sentences
- Speech and audio
- Stock prices
- Weather records
- Daily sales data

## 2. Why do we need RNN?
Traditional neural networks generally process each input independently, or require a fixed-size representation of the input. They do not naturally maintain a hidden state across a sequence.

**RNNs are designed to handle this sequential relationship**

### Traditional Neural Network
- Usually processes fixed-format inputs
- No built-in recurrent memory
- Does not naturally pass information across time steps
- Useful for many classification tasks

### Recurrent Neural Network
- Designed for sequential inputs
- Maintains a hidden state
- Passes information from one time step to the next
- Useful for text, speech, and time-series tasks

### Example:
Suppose we want to understand the sentence:
"I went to the bank to deposit money."
The word "money" helps clarify that "bank" refers to a financial institution. RNNs can use information from earlier words to help interpret later words.

## Why do we need RNN technically?

A Recurrent Neural Network (RNN) is a neural network designed to process sequential data by maintaining a hidden state that carries information from previous time steps.

### 1. The problem with a normal ANN

A traditional feedforward ANN processes input features without explicitly maintaining information from previous time steps.

For example, in a sentence:

“I grew up in India, and I speak fluent Hindi.”

To understand the context of the last word, the model may need information from earlier words. A basic ANN doesn't naturally maintain this sequential context.

### 2. How RNN solves this problem

RNN processes data one time step at a time and passes its hidden state to the next step.

\\(x_1\\)

Word 1

\\(h_1\\)

Hidden state

\\(x_2\\)

Word 2

The hidden state carries information forward as the sequence is processed.

### 3. The mathematical formula

The hidden state is calculated as:

\\[ h_t=\tanh(W_xx_t+W_hh\_{t-1}+b_h) \\]

Output is calculated as:

\\[ y_t=f(W_yh_t+b_y) \\]

Where:

- \\(x_t\\): input at the current time step.
- \\(h_t\\): current hidden state (learned representation of the sequence so far).
- \\(h\_{t-1}\\): previous hidden state.
- \\(W_x,W_h,W_y\\): learnable weight matrices.
- \\(b_h,b_y\\): biases.
- \\(\tanh\\): activation function that helps introduce non-linearity.

Key point: The same weights are reused across time steps, allowing the RNN to learn sequential patterns.

### 4. Why is this useful?

| Problem                 | Why RNN helps                              |
| ----------------------- | ------------------------------------------ |
| Text prediction         | Uses preceding words as context            |
| Time-series forecasting | Learns patterns from previous observations |
| Speech recognition      | Processes audio features in sequence       |
| Sentiment analysis      | Processes words in their order             |

Limitation: Basic RNNs can suffer from vanishing gradients, making long-term dependencies difficult to learn. LSTM and GRU architectures help address this problem.

Bhai, RNN ke 4 components ko ek simple example se samjho.

Imagine you're reading: "I love eating \_\_\_"

1\. Input (\\(x_t\\)): Current word jo RNN ko diya jata hai. Example: `"eating"`.

2\. Hidden State (\\(h_t\\)): RNN ki memory jaisi hoti hai. Ye previous words `"I love"` se mili information ko current word `"eating"` ke saath combine karti hai.

3\. Recurrent Connection: Ye previous step ki memory ko next step tak pahunchata hai, taaki RNN context na bhule.

4\. Output (\\(y_t\\)): RNN ki prediction. Example: `"food"` ya `"pizza"`.

Easy trick to remember:

- Input = Nayi information
- Hidden State = Information ki memory
- Recurrent Connection = Memory ko aage bhejna
- Output = Prediction

Yahi process har time step par repeat hota hai, jisse RNN sequential data ko samajh sakta hai.

Bhai, RNNs ke 4 main types hote hain, based on input aur output sequence ke structure par.

## 1. One-to-One

One input → One output

Input

Output

- Example: Image classification.
- Note: Ye traditional RNN ka typical use case nahi hai; simple feedforward neural network bhi use ho sakta hai.

## 2. One-to-Many

One input → Multiple outputs

One input

Output 1 → Output 2 → Output 3

- Example: Image caption generation.
- Ek image se multiple words ki caption generate karna.

## 3. Many-to-One

Multiple inputs → One output

Input 1 → Input 2 → Input 3

One output

- Example: Sentiment analysis.
- Sentence ke saare words process karke decide karna ki sentiment positive hai ya negative.

## 4. Many-to-Many

Multiple inputs → Multiple outputs

Input 1 → Input 2 → Input 3

Output 1 → Output 2 → Output 3

- Example: Video frame labeling, sequence tagging, and machine translation.
- Multiple inputs ko process karke multiple outputs generate karna.

## Important: RNN architectures

Ye input-output types se alag classification hai. Common recurrent architectures hain:

| Architecture      | Main feature                                                                 |
| ----------------- | ---------------------------------------------------------------------------- |
| Vanilla RNN       | Basic recurrent network                                                      |
| LSTM              | Long-term information learn karne ke liye gates aur cell state use karta hai |
| GRU               | LSTM se simpler gated architecture                                           |
| Bidirectional RNN | Sequence ko forward aur backward dono directions mein process karta hai      |
| Stacked RNN       | Multiple recurrent layers ko ek ke upar ek rakhta hai                        |

Interview tip: One-to-one, one-to-many, many-to-one, aur many-to-many input-output patterns hain; LSTM, GRU, aur Bidirectional RNN architectures hain.

```python
# Step 1: Importing
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Step 2: Define the input text
text = 'A Recurrent Neural Network (RNN) is a class of artificial neural networks designed to process sequential data and time-series information by maintaining a "hidden state" (memory) that passes from one step to the next'
chars = sorted(list(set(text)))
char_to_index = {char: i for i, char in enumerate(chars)}
# FIXED: Added missing index_to_char lookup dictionary
index_to_char = {i: char for i, char in enumerate(chars)} 

# Step 3: Create Sequences and Labels
seq_length = 3
sequences = []
labels = []

for i in range(len(text) - seq_length):
    seq = text[i:i + seq_length]
    label = text[i + seq_length]
    sequences.append([char_to_index[char] for char in seq])
    labels.append(char_to_index[label])

x = np.array(sequences)
y = np.array(labels)

# Step 4: Convert Sequences and Labels to One-Hot Encoding
X_one_hot = tf.one_hot(x, len(chars))
Y_one_hot = tf.one_hot(y, len(chars))

# Step 5: Build the Model
model = Sequential()
model.add(SimpleRNN(50, input_shape=(seq_length, len(chars)), activation='relu'))
model.add(Dense(len(chars), activation='softmax'))

# Step 6: Compile and Train the Model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_one_hot, Y_one_hot, epochs=100)

# Step 7: Generate New Text Using the Trained Model
start_seq = "A Recurrent Neural Network (RNN)"
generated_text = start_seq

for i in range(50):
    x_input = np.array([[char_to_index[char] for char in generated_text[-seq_length:]]])
    x_one_hot = tf.one_hot(x_input, len(chars))
    # FIXED: Added verbose=0 to prevent prediction output from cluttering the terminal
    prediction = model.predict(x_one_hot, verbose=0) 
    next_index = np.argmax(prediction)
    next_char = index_to_char[next_index]
    generated_text += next_char

print("\nGenerated Text:")
print(generated_text)
```
