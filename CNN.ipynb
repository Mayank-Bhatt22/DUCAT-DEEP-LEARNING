# What is CNN (Convolutional Neural Networks)
**Convolutional Neural Networks (CNNs)** are deep learning models designed to process data with a grid-like topology such as images. They are the foundation for most modern computer vision applications to detect features within visual data.

# Working of CNN
1. **Input Image:** CNN receives an input image which is preprocessed to ensure uniformity in size and format.
2. **Convolutional Layers:** Filters are applied to the input image to extract features like edges, textures and shapes.
3. **Pooling Layers:** The feature maps generated by the convolutional layers are downsampled to reduce dimensionality.
4. **Fully Connected Layers:** The downsampled feature maps are passed through fully connected layers to produce the final output, such as a classification label.
5. **Output:** The CNN outputs a prediction, such as the class of the image.

# Core Layers of a CNN
- **Input Layer:** Receives the digital image as a grid of numbers representing pixel colors and brightness values.
- **Convolutional Layer:** Uses a small grid of numbers called a kernel or filter that slides across the image. It multiplies filter values with local pixel values and adds them up to create a feature map that highlights specific patterns like vertical lines or curves.
- **ReLU (Rectified Linear Unit) Layer:** Applies a mathematical operation that turns negative values into zero, adding non-linearity so the network can learn complex patterns.
- **Pooling Layer:** Shrinks the size of the feature map (often by taking the maximum value in a small window, called max pooling), which reduces the needed computer power and keeps the most important data.
- **Flattening Layer:** Transforms the final 2D feature maps into a 1D column of numbers so it can fit into standard neural network layers.
- **Fully Connected Layer:** Uses the extracted features to score and classify the image into final categories (such as identifying a cat, car, or stop sign).

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    Flatten, Dense
)

# Load dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize images
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
X_train = X_train[..., None]
X_test = X_test[..., None]

# Build CNN
model = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(
        32,
        (3, 3),
        activation="relu"
    ),

    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(10, activation="softmax")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train model
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

# Evaluate model
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

# Explanation of the Code

- **`Conv2D(32, (3,3))`** → Applies 32 filters of size 3×3 to detect features in the image.
- **`activation='relu'`** → Adds non-linearity and helps the model learn complex patterns.
- **`MaxPooling2D((2,2))`** → Reduces the feature-map dimensions while retaining important features.
- **`Flatten()`** → Converts the 2D feature maps into a 1D vector.
- **`Dense(128)`** → Uses 128 neurons to learn higher-level patterns.
- **`Dense(10)`** → Predicts one of the 10 digit classes (0–9).
- **`softmax`** → Converts the outputs into probabilities for each class.
- **`adam`** → Optimizes the model's weights during training.
- **`epochs=5`** → Trains the model for 5 complete passes through the training data.

# Important CNN Terms
- `Filter/Kernel:` A small matrix used to extract image features.
- `Feature Map:` The output produced by applying a filter.
- `Stride:` The number of pixels a filter moves at each step.
- `Padding:` Extra pixels added around an image to control output dimensions.
- `Epoch:` One complete pass through the training dataset.
- `Batch Size:` Number of training examples processed together
